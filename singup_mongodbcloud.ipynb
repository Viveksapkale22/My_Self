```python
from pymongo import MongoClient
import bcrypt

# Correct MongoDB URI
MONGO_URI = "mongodb+srv://cluster0.mcjuw.mongodb.net/?authSource=%24external&authMechanism=MONGODB-X509&retryWrites=true&w=majority&appName=Cluster0"

# Connect to MongoDB using X.509 authentication
client = MongoClient(
    MONGO_URI,
    tls=True,
    tlsCertificateKeyFile="/content/X509-cert-7398551624606348947.pem"
)

# Select Database & Collection
db = client["UserDB"]
users_collection = db["users"]

# Hash Password Function
def hash_password(password):
    return bcrypt.hashpw(password.encode('utf-8'), bcrypt.gensalt()).decode('utf-8')

# Signup Function
def signup(username, email, password):
    if users_collection.find_one({"email": email}):
        return "Email already exists. Try logging in."

    hashed_password = hash_password(password)
    user_data = {"username": username, "email": email, "password": hashed_password}
    users_collection.insert_one(user_data)
    return "Signup successful!"

# Get User Input
username = input("Enter username: ")
email = input("Enter email: ")
password = input("Enter password: ")

# Insert Data
print(signup(username, email, password))
```
